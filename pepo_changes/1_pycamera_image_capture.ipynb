```python
from picamera2 import Picamera2, Preview
```

```python
# Create one object per camera, otherwise will yield resource busy error.
picam2 = Picamera2()
```

```python
# camera tutorial: https://www.tomshardware.com/how-to/use-picamera2-take-photos-with-raspberry-pi
camera_config = picam2.create_still_configuration(main={"size": (1920, 1080)}, lores={"size": (640, 480)}, display="lores")
```

```python
picam2.configure(camera_config)
```

```python
# PAGE 45 from picamera2 library documentation.
# picamera2-manual.pdf
picam2.start_and_capture_file("./test_images/vs_code_test.jpg")
```
